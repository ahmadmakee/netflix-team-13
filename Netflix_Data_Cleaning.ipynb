# Netflix Data Analysis – Challenge 1
## Data Cleaning and Preparation
This section prepares the Netflix datasets for analysis. We will check missing values, duplicates, data types,
ratings, dates, and data consistency.

```python
import pandas as pd

movies = pd.read_csv("movies.csv")
users = pd.read_csv("users.csv")
watch_history = pd.read_csv("watchhistory.csv")
netflix_results = pd.read_csv("netflix_results.csv")

print("All four datasets loaded successfully!")
```

```python
datasets = {
    "Movies": movies,
    "Users": users,
    "Watch History": watch_history,
    "Netflix Results": netflix_results
}

for name, df in datasets.items():
    print(f"\n===== {name} =====")
    print("Rows and columns:", df.shape)
    print("Missing values:", df.isna().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())
```

```python
for name, df in datasets.items():
    print(f"\n===== {name} =====")
    print(df.dtypes)
```

```python
# Since dates showed as texts, we need to convert them into proper date formats
users["date_of_birth"] = pd.to_datetime(
    users["date_of_birth"], errors="coerce"
)

watch_history["watch_date"] = pd.to_datetime(
    watch_history["watch_date"], errors="coerce"
)

netflix_results["date_of_birth"] = pd.to_datetime(
    netflix_results["date_of_birth"], errors="coerce"
)

netflix_results["watch_date"] = pd.to_datetime(
    netflix_results["watch_date"], errors="coerce"
)

# Check the results
print("Users:")
print(users.dtypes)

print("\nWatch History:")
print(watch_history.dtypes)

print("\nNetflix Results:")
print(netflix_results.dtypes)

# Check whether any dates failed to convert
for name, df in [
    ("Users", users),
    ("Watch History", watch_history),
    ("Netflix Results", netflix_results)
]:
    for column in ["date_of_birth", "watch_date"]:
        if column in df.columns:
            print(f"{name} - {column}: {df[column].isna().sum()} invalid/missing dates")
```

```python
# Next is to validate ratings, IDs and release years
from datetime import datetime

print("Invalid ratings:",
      (~watch_history["rating"].between(1, 5)).sum())

print("Duplicate movie IDs:",
      movies["movie_id"].duplicated().sum())

print("Duplicate user IDs:",
      users["user_id"].duplicated().sum())

print("Unknown movie IDs:",
      (~watch_history["movie_id"].isin(movies["movie_id"])).sum())

print("Unknown user IDs:",
      (~watch_history["user_id"].isin(users["user_id"])).sum())

print("Invalid release years:",
      (~movies["release_year"].between(
          1888, datetime.now().year
      )).sum())
```

```python
netflix_results.to_csv("cleaned_netflix_results.csv", index=False)

print("Cleaned Netflix dataset saved successfully!")
```
